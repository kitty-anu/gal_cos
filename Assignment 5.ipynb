### Q3: The Super-neutrino strikes! (25%)

In the year 2017, the IceCube detector registered a neutrino with an energy of 290 TeV that was attributed to an outburst in an AGN called TXS 0506+056. The proper distance to this AGN has now been measured to be 4 billion light years.

*Note: throughout this question you will be dealing with very small numbers. Please either use high precision code or appropriate approximations to deal with them; the physics is in these very small differences!*

i) Assuming a rest mass of 0.1 eV for neutrinos, **calculate** the velocity of the neutrino relative to the vacuum speed of light.

```python
import numpy as np
from scipy.constants import c, e, epsilon_0, m_e, pi

# Constants
c_light = c  # speed of light in m/s
m_nu = 0.1 * 1.60218e-19 / (c_light**2)  # neutrino rest mass (0.1 eV/c^2) in kg
E_nu = 290 * 1.60218e-7  # neutrino energy (290 TeV) in joules
distance_ly = 4e9  # distance in light-years
distance_m = distance_ly * 9.461e15  # converting distance to meters

# Neutrino velocity using the relativistic equation
v_neutrino = c_light * np.sqrt(1 - (m_nu * c_light**2 / E_nu)**2)
print(f"Neutrino velocity: {v_neutrino:.10f} m/s")
```

ii) Assuming space between the AGN and Earth was a perfect vacuum, **calculate** how much later the neutrino would arrive after simultaneously emitted photons.

```python
# Time taken by photons (c_light) and neutrinos (v_neutrino) to cover the distance
t_photon = distance_m / c_light
t_neutrino = distance_m / v_neutrino

# Time delay
delta_t = t_neutrino - t_photon
print(f"Time delay between neutrino and photon: {delta_t:.10f} seconds")
```

iii) In reality, even intergalactic space is not a perfect vacuum, but filled with low-density plasma that increases the refractive index of the intergalactic medium. Assuming intergalactic space has an electron density of 1 per m³, **calculate** by how much the arrival of UV light with λ = 300 nm from TXS 0506+056 is delayed as a result.

```python
# Electron density and UV wavelength
electron_density = 1  # electrons per m^3
lambda_uv = 300e-9  # UV wavelength in meters

# Calculating refractive index
n = 1 + (e**2 / (2 * pi * m_e * epsilon_0 * c_light**2)) * (electron_density / lambda_uv**2)
v_uv = c_light / n  # Speed of UV light in plasma
print(f"Refractive index for UV light: {n:.10f}")

# Time delay due to refractive index
t_uv = distance_m / v_uv
delta_t_uv = t_uv - t_photon
print(f"UV light delay due to plasma: {delta_t_uv:.10f} seconds")
```

iv) **Comment** on what arrives first, the neutrino mentioned above, or the UV-blue light.

```python
if delta_t < delta_t_uv:
    print("The neutrino arrives first.")
else:
    print("The UV light arrives first.")
```

v) The AGN also emitted radio waves alongside all the other radiation and particles. If a 100 MHz radio wave was emitted at the same time as the UV light in (iii), **calculate** how far away the radio wave would be from the Earth the moment the UV light arrives.

```python
# Frequency of the radio wave
f_radio = 100e6  # 100 MHz
lambda_radio = c_light / f_radio  # Wavelength of the radio wave

# Distance the radio wave travels in the time UV light is delayed
distance_radio = lambda_radio * (delta_t_uv / (1/f_radio))
print(f"Distance traveled by radio wave: {distance_radio:.10f} meters")
```
